```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Revisar rangos de cada columna numérica
print(df.describe())

# 2. Detectar valores únicos en columnas categóricas (para ver si hay categorías raras)
for col in df.select_dtypes(include="object").columns:
    print(f"{col}: {df[col].unique()}")

# 3. Boxplots para detectar outliers en variables numéricas
for col in df.select_dtypes(include="number").columns:
    plt.figure(figsize=(6,3))
    sns.boxplot(x=df[col])
    plt.title(f"Outliers en {col}")
    plt.show()

# 4. Conteo de valores en columnas binarias (para confirmar solo 0 y 1)
for col in df.columns:
    if set(df[col].unique()) <= {0,1}:
        print(f"{col}: {df[col].value_counts().to_dict()}")
```

```python
# Ejemplo de chequeo de outliers en variables clave
print(df[df["age"] < 10])              # edades sospechosas
print(df[df["creatinine"] > 20])       # creatinina fuera de rango clínico
print(df[df["glucose"] > 400])         # glucosa muy alta
print(df[df["urea"] > 200])            # urea muy alta
```
